```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/content/Startup.csv")

data["year"] = pd.to_numeric(data["year"], errors="coerce")
data["amount"] = pd.to_numeric(data["amount"], errors="coerce")

data = data.dropna(subset=["year", "amount"])

investment_per_year = data.groupby("year")["amount"].sum().reset_index()

plt.figure(figsize=(10, 6))

plt.plot(
    investment_per_year["year"],
    investment_per_year["amount"],
    marker="o",
    linewidth=2,
    label="Total Investment"
)

plt.title("Total Investment by Year", fontsize=18)
plt.xlabel("Year", fontsize=14)
plt.ylabel("Investment Amount", fontsize=14)
plt.grid(True, linestyle="--")
plt.legend()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/content/Startup.csv")

data["year"] = pd.to_numeric(data["year"], errors="coerce")
data["amount"] = pd.to_numeric(data["amount"], errors="coerce")

data = data.dropna(subset=["year", "amount"])

investment_per_year = data.groupby("year")["amount"].sum().reset_index()

plt.figure(figsize=(10, 6))

plt.plot(
    investment_per_year["year"],
    investment_per_year["amount"],
    marker="o",
    markersize=8,
    linestyle="--",
    linewidth=3,
    label="Total Amount"
)

plt.title("Year Wise Amount Investment", color="yellow", fontsize=25)

plt.xlabel("Year", fontsize=14)
plt.ylabel("Investment Amount", fontsize=14)

plt.grid(True, linestyle="--")
plt.legend()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/content/Startup.csv")

data["city"] = data["city"].astype(str).str.strip()

city_startups = data["city"].value_counts().head(10)

plt.figure(figsize=(10, 6))

plt.bar(
    city_startups.index,
    city_startups.values,
    color="skyblue"
)

plt.gca().set_facecolor("lightyellow")
plt.gcf().set_facecolor("lightgray")

plt.title("Top 10 Cities Highest Number of Startups", color="red", fontsize=20)
plt.xlabel("Cities", fontsize=14)
plt.ylabel("Number of Startups", fontsize=14)

plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("/content/Startup.csv")

data["round"] = data["round"].astype(str).str.strip()

round_distribution = data["round"].value_counts().head(10)

plt.figure(figsize=(12, 6))

bars = plt.bar(
    round_distribution.index,
    round_distribution.values,
    color=[
        "red",
        "blue",
        "green",
        "orange",
        "purple",
        "yellow",
        "pink",
        "cyan",
        "brown",
        "gray"
    ]
)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        str(int(bar.get_height())),
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.title("Distribution of Funding Rounds", color="red", fontsize=22)
plt.xlabel("Funding Rounds", fontsize=14)
plt.ylabel("Number of Startups", fontsize=14)

plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Now are startup funding amounts distrubuted
import pandas as pd
import matplotlib.pyplot as plt

# Read the dataset
data = pd.read_csv("/content/Startup.csv")

# Convert amount column to numeric
data["amount"] = pd.to_numeric(data["amount"], errors="coerce")

# Remove missing values
data = data.dropna(subset=["amount"])

# Create Histogram
plt.figure(figsize=(12, 6))

plt.hist(
    data["amount"],
    bins=25,
    color="skyblue",
    edgecolor="black",
    linewidth=1.2
)

plt.title("Distribution of Startup Funding Amounts", color="red", fontsize=22)
plt.xlabel("Funding Amount", fontsize=14)
plt.ylabel("Frequency", fontsize=14)

plt.grid(axis="y", linestyle="--", alpha=0.6)

plt.show()
```

```python
# The Realtionship between Funding Your and Funding Amount?
plt.scatter(data["year"],data["amount"])
plt.title("Funding Amount vs Funding Year")
plt.xlabel("Funding Year")
plt.ylabel("Funding Amount")
plt.show()
```
